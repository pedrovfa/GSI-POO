## <center> Funções 🎼 </center>

### Funções
- Conjunto de pares ordenado de entrada (**domínio**) e saída (**contra-domínio**)
    - Definição por **extensão** ou por **abstração**
    - Como utilizar funções? **Definição** e **Aplicação**

<small>

```
            f(x, y) -> x + y             Definição por abstração

            f(3, 5) -> 8                Aplicação!
```

</small>

- Em **Programação**, a utilização de funções também:
    - compreende a **definição da função** (nome, parâmetros e algoritmo) 
    - e a **aplicação da função** (execução do algoritmo sobre valores passados como argumentos).
<br><br>

- Exemplo de funções em C# que já conhecemos: `Console.WriteLine(...)`, `Console.ReadLine()`, `Convert.ToInt32(...)`
<br><br>





## <center>Definição de uma [função em C#](https://docs.microsoft.com/en-us/dotnet/csharp/programming-guide/classes-and-structs/local-functions) </center>

```csharp
            <tipo-valor-retorno> <nome-da-funcao>(<tipo-parametro-1> <nome-parametro-1>, <tipo-parametro-2> <parametro-2>, ...) {
                // corpo da função
            }
```

- Temos sempre que indicar o **tipo de dados que a função retorna** `<tipo-valor-retorno>`. O tipo [`void`](https://docs.microsoft.com/en-us/dotnet/csharp/language-reference/builtin-types/void) indica que não tem retorno - neste caso designamos de **procedimento** em vez de função.
<br><br>

- Temos que **dar um nome à função** `<nome-da-funcao>`
<br><br>

- Temos que **definir quais os parâmetros formais indicando o tipo de dados e o nome** `<tipo-parametro> <nome-parametro>` separados por `,`. **A função pode ter 0 ou mais parâmetros**.
<br><br>

- **O corpo da função representa o algoritmo** para o cálculo do valor da função. **A instrução [`return <expressão>;`](https://docs.microsoft.com/en-us/dotnet/csharp/language-reference/statements/jump-statements#the-return-statement) termina a execução da função e o valor avaliado da `<expressão>` será o resultado da função**. No caso de um procedimento (tipo do retorno definido como `void`) não é necessário usar o `return`, mas se o usarmos terá de ser no formato `return;` que termina igualmente a execução do procedimento.
<br><br>

- Exemplos

```csharp
        int Sum(int n1, int n2) {
            return n1 + n2;
        }

        int Square(int x) {
            return x * x;
        }

        void Greetings(string username) {
            Console.WriteLine($"Hi {username}! Welcome to OOP");
        }

```

- Se o **corpo da função for apenas o return de uma avalição de uma expressão** podemos usar uma forma mais reduzida para o corpo da funcção, omitindo o bloco `{}` e o `return`:

```csharp
        double Price(double cost) => cost * 1.23;
```


````polyglot-notebook
int Sum(int n1, int n2) {
    return n1 + n2;
}

int Square(int x) {
    return x * x;
}

void Greetings(string username) {
    Console.WriteLine($"Hi {username}. Welcome to upskill!");
}
````

## <center>Aplicação de Funções em C# </center>


```csharp
                <nome-da-funcao>(<argumento-1>, <argumento-2>,...)
```

- A aplicação de uma função é tipicamente designada por **chamar uma função**, ou **invocar uma função**.
<br><br>

- Após termos definido a função, passa a existir uma nova **expressão correspondente à aplicação da função**
<br><br>

- **Os argumentos da aplicação da função** são os valores que são passados para os parâmetros formais da função. Os tipos de dados dos argumentos têm que ser compatíveis com os tipos definidos no parâmetro (conversão implícita ou explícita).
<br><br>

## <center>Exemplo de Definição e Aplicação de Função</center>

- Exemplo: Função para calcular uma progressão aritmética

```csharp 
    // Progressao Aritmetica de N:  1 + 2 + 3 + ... + N 
    int SomaProgressaoAritmetica(int n) {
        int soma = 0;
        int iter = 1;
        while (iter <= n) {
            soma = soma + iter;
            iter = iter + 1;
        }
        return soma;
    }

    int soma = 4 + 6;
    Console.WriteLine(SomaProgressaoAritmetica(100));
    Console.WriteLine(soma);
```

- **O que acontece com a variável soma!?**



````polyglot-notebook
int SomaProgressaoAritmetica(int n) {
    int soma = 0;
    int iter = 1;
    while (iter <= n) {
        soma = soma + iter;
        iter = iter + 1;
    }
    return soma;
}

int soma = 4 + 6;
Console.WriteLine(SomaProgressaoAritmetica(100));
Console.WriteLine(soma);
````

## <center>Abstração Procedimental</center>

- A criação de uma **função corresponde a uma forma de abstração**: nomear uma sequência de operações que permitem atingir um determinado objetivo
<br><br>

- As funções permitem aos programadores pensar no **que** (faz a função) e não no **como** (a funcão é implementada)
<br><br>

- Podemos considerar **as funções como caixas pretas**

```csharp
    int SomaProgressaoAritmetica(int n) {
        int soma = 0;
        int iter = 1;
        while (iter <= n) {
            soma = soma + iter;
            iter = iter + 1;
        }
        return soma;
    }
```

- Será que existe uma solução não iterativa (sem ciclos) para a progressão aritmética? 

```csharp
    int SomaProgressaoAritmetica(int n) {
        if (n < 1)
            return 0;

        return n * (n + 1) / 2;
    }

```

## <center> Métodos de Passagem de Parâmetros </center>

- Vimos que quando uma função é chamada estabelece-se uma **correspondência entre os parâmetros da função e os argumentos da chamada**. O processo de ligação entre os parâmetros é denominado de **método de passagem de parâmetros**.

- Existem vários métodos de passagem de parâmetros, e cada linguagem de programação utiliza um ou vários destes métodos para a comunicação com funções.
<br><br>

- Em C# os dois métodos mais usados são:
    - **Passagem por valor**
    - **Passagem por referência**


## <center>Passagem por Valor <center>

- Um parâmetro em que seja utilizado a **passagem por valor** comporta-se, dentro da sua função, como um nome local que é inicializado no início da avaliação da função, **funciona como uma cópia**

- Exemplo:

```csharp
        void Troca(int x, int y) {
            Console.WriteLine($"Antes da troca: x={x}, y={y}");
            // Troca valor dos parâmetros recorrendo a uma variavel temporária
            // Alternativamente podiamos ter feito simplesmente (x, y) = (y, x);
            int temp = x;
            x = y;
            y = temp;
            Console.WriteLine($"Depois da troca: x={x}, y={y}");
        }

        int x = 3;
        int y = 10;
        Troca(x, y);
        Console.WriteLine($"x={x}");
        Console.WriteLine($"y={y}");
```



## <center>Passagem por Referência ([ref](https://docs.microsoft.com/en-us/dotnet/csharp/language-reference/keywords/ref) ou [out](https://docs.microsoft.com/en-us/dotnet/csharp/language-reference/keywords/out-parameter-modifier))  <center>

- Um parâmetro em que seja utilizado a **passagem por referência**, o que é associado ao parâmetro não é o valor do argumento, mas sim a **localização na memoria do computador que contém o seu valor**
- Qualquer **modificação feita no parâmetro dentro da função, reflete-se no argumento original** (o qual tem que ser uma variável)

- Exemplo:

```csharp
        void Troca(ref int x, ref int y) {
            Console.WriteLine($"Antes da troca: x={x}, y={y}");
            // Troca valor dos parâmetros recorrendo a uma variavel temporária
            // Alternativamente podiamos ter feito simplesmente (x, y) = (y, x);
            int temp = x;
            x = y;
            y = temp;
            Console.WriteLine($"Depois da troca: x={x}, y={y}");
        }

        int x = 3;
        int y = 10;
        Troca(ref x, ref y);
        Console.WriteLine($"x={x}");
        Console.WriteLine($"y={y}");
```


- Os **modificadores `ref` ou `out` têm como consequência a passagem de argumentos por referência**. 
    - O **`ref`** obriga a que o argumento tenha sido inicializado antes de ser aplicado na função - **funciona como argumento de entrada e saída**
    - O **`out`** não obriga a inicializar antes da chamada - **funciona apenas como argumento de saída** (e tem que ser assignado obrigatoriamente na definição da função).

```csharp
        int numero;
        while (!int.TryParse(Console.ReadLine(), out numero)) {
            Console.WriteLine("Número inválido. Tente novamente");
        }

        Console.WriteLine($"O numero introduzido foi: {numero}");
```

## <center> [Argumentos opcionais](https://docs.microsoft.com/en-us/dotnet/csharp/programming-guide/classes-and-structs/named-and-optional-arguments#optional-arguments) </center>

- Quando definimos a função **podemos indicar a existência de parâmetros opcionais, os quais assumem um valor default**, caso não sejam passados como argumento.
<br><br>

- Exemplo: validação genérica da idade com base 

```csharp

    // minAge e maxAge são parâmetros opcionais com valores default
    // de 0 e 130 respetivamente
    bool IsValidAge(int age, int minAge = 0, int maxAge = 130) {
        return age >= minAge && age <= maxAge;
    }

    Console.WriteLine($"Idade Válida (16) -> {IsValidAge(16)}");
    Console.WriteLine($"Idade Válida (25, 55) -> {IsValidAge(25, 55)}");
    Console.WriteLine($"Idade Válida (16, 0, 12) -> {IsValidAge(16, 0, 12)}");

```

- Então mas se quiser apenas especificar a idade máxima limite, ex: 12, ignorando a mínima, tenho que também passar a mínima? Não! Nesses casos **podemos dizer exatamente qual o argumento que estamos a passar**, colocando o `<nome-do argumento>:` antes do seu valor.

- Exemplo:
    
```csharp    
        bool ageOk = IsValidAge(16, maxAge: 12);
```

- **Nalgumas situações** de argumentos opcionais, **a utilização dos nomes nos argumentos, ajuda imenso a leitura do código**.


## <center> Módulos (_Libraries_) </center>

- A maior parte das linguagens de programação suportam a abordagem de **Programação Modular**
  - técnica de desenvolvimento de software que permite **separar a funcionalidade de um programa em vários módulos**
  - um módulo é tipicamente definido num ou mais ficheiros separados
<br><br>  

- Um **módulo**, também conhecido como **biblioteca (_library_)** pode ser visto como uma **coleção de funções agrupadas numa ou mais entidades**, e que visa resolver uma parte do problema.

- Segundo esta definição, podemos dizer que em **C# já interagimos com diversos módulos**:
    - Vimos anteriormente que em C# podiamos usar a **entidade `Math`** a qual contem inúmeras funções matemáticas (`Round`, `Sqrt`, `Sin`, etc.)
    - Temos usado também a **entidade `Console`** para interagir com a consola, que contém inúmeras funções (`WriteLine`, `ReadLine`, etc.)
<br><br>

- Em C# é **mais comum ouvirmos frases como _"estou a usar a classe Math"_ do que _"estou a usar o módulo Math"_**, (vêr nota abaixo)

<small>

**_Nota_**: Em C# as entidades `Math` e `Console` na realidade são **classes**, e as suas funções são na realidade **métodos**. Veremos mais à frente esta terminologia quando visitarmos a temática da Programação Orientada por Objetos.

</small>

## <center> C# - Módulos (_Libraries_) e _Namespaces_</center>

- Em C#, **as entidades encontram-se estruturadas e organizadas por _namespaces_**
    - As entidades `Math` e `Console` encontra-se no **namespace `System`**
    - Para usarmos estas entidades temos que, ou especificar o nome completo
        - `System.Math.Round(25.36)`
    - Ou alternativamente, temos que **importar o namespace `System` através da diretiva `using` (no topo do ficheiro)**

```csharp
        using System;

        Console.WriteLine(Math.Round(25.36));
```

- Mas nos nossos programas não temos visto o `using System;` **Porquê**?
    - Porque **temos usado os usings implícitos do C# 10**, onde alguns usings são automaticamente importados, tais como o `using System;`
<br><br>

- Vamos supor que queremos **ler o conteúdo de um ficheiro**. Para tal podemos usar a entidade `File`, em particular a função `ReadAllText`. 
    - **A entidade `File` encontra-se no namespace `System.IO`**, pelo que **temos que o importar** para poder usar o `File`.

```csharp

        var fileContent = File.ReadAllText("text.txt");
        Console.WriteLine(fileContent);
            
```

- No exemplo acima não estamos a importar o `System.IO`, pelo que **vai dar erro de compilação** pois a entidade `File` não está a ser reconhecida. Teriamos que fazer o seguinte

```csharp
        using System.IO;
        
        var fileContent = File.ReadAllText("text.txt");
        Console.WriteLine(fileContent);
```




````polyglot-notebook
using System;
using System.IO;

var fileContent = File.ReadAllText("text.txt");
Console.WriteLine(fileContent);
````
